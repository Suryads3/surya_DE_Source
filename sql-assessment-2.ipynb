```python
%%sql
select employee_name, department, salary, lag(salary, 1) over(partition by department order by salary) as previous_sal from employees;
```

```python
%%sql
select employee_name, department, salary, previous_sal, (previous_sal-salary) as salary_diff from (
select employee_name, department, salary, lag(salary, 1) over(partition by department order by salary) as previous_sal from employees) t;
```

```python
%%sql
select employee_name, department, salary, previous_sal from (
select employee_name, department, salary,
lag(salary) over(partition by department order by salary) as previous_sal
from employees
)t
where salary > previous_sal;
```

```python
%%sql
select employee_name, department, salary, next_emp_sal from (
select employee_name, department, salary,
LEAD(salary,1) over(partition by department order by salary) as next_emp_sal
from employees
)t
```

```python
%%sql
select employee_name, department, salary, previous_sal, next_emp_sal from (
    select employee_name, department, salary, previous_sal, 
    LEAD(salary,1, 0) over(partition by department order by salary) as next_emp_sal from (
    select employee_name, department, salary,
        LAG(salary,1,0) over(partition by department order by salary) as previous_sal 
from employees 
) t1
) t2
where (salary > previous_sal) and (salary > next_emp_sal)
```

```python
%%sql
with employee_salary as (
    select employee_name, department, salary, avg(salary) over(partition by department order by salary) as avg_sal from employees
)
select employee_name, department, salary, avg_sal from employee_salary
```

```python
%%sql
with employee_salary as (
    select employee_name, department, salary, ROW_NUMBER() over(partition by department order by salary) as high_paid_sal from employees
)
select employee_name, department, salary, high_paid_sal from employee_salary
```

```python
%%sql
with employee_salary as (
    select employee_name, department, salary, ROW_NUMBER() over(partition by department order by salary) as high_paid_sal from employees
) 
select employee_name, department, salary, high_paid_sal from employee_salary where high_paid_sal =2
```

```python
%%sql
with employee_salary as (
    select employee_name, department, salary, DENSE_RANK() over(partition by department order by salary) as high_paid_sal from employees
) 
select employee_name, department, salary, high_paid_sal from employee_salary where high_paid_sal =3
```

```python
%%sql
with employee_salary as (
    select employee_name, department, salary, coalesce(bonus,0) as bonus, (salary + bonus) as total_income, 
    dense_rank() over(partition by department order by salary) as d_rank from employees
) 
select employee_name, department, salary, bonus, total_income, d_rank from employee_salary where d_rank <=5
```

```python
%%sql
select employee_name, department, salary, performance_score, perform_rank from (
select employee_name, department, salary, performance_score,   
DENSE_RANK() over(partition by department order by performance_score desc, salary asc) as perform_rank from employees
) t
where perform_rank <=2
```

```python
%%sql
select employee_name, department, salary, performance_score, avg_perform_score from (
select employee_name, department, salary, performance_score,
AVG(performance_score) over(partition by department) as avg_perform_score from employees
) t
where performance_score > avg_perform_score
```

```python
%%sql
with department_salary_val as (
    select employee_name, department, salary, 
    sum(salary) over(partition by department order by salary) as department_salary
    from employees
    
),
calculate_employee_data as (
    select employee_name, department, salary, department_salary,
    (salary/department_salary*100) as salary_percentage from department_salary_val
    
)

select *
from calculate_employee_data where salary_percentage > 10;
```
