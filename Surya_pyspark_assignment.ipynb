```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
spark = SparkSession.builder.appName("rdd_df_pratice").getOrCreate()
sc = spark.sparkContext

numbers = [5, 12, 7, 20, 3, 16, 9, 24, 10, 1]
rdd= sc.parallelize(numbers)
rdd_filter_one = rdd.filter(lambda num: num > 10)
rdd_filter_two = rdd_filter_one.map(lambda num: num * 2)
rdd_filter_two.collect()
```

```python
student_marks = [
    ("Arun", 78),
    ("Divya", 92),
    ("Kavin", 45),
    ("Meena", 66),
    ("Ravi", 35),
    ("Priya", 88)
]
rdd_std = sc.parallelize(student_marks)
rdd_filter_mark = rdd_std.filter(lambda marks: marks[1] >=50)
rdd_filter_mark.collect()
rdd_show_data = rdd_filter_mark.map(lambda filtered_marks: str(filtered_marks[0]) + " scored "+ str(filtered_marks[1]) )
rdd_show_data.collect()

# rdd_std_mark_filter = rdd_std.filter(lambda num: num >= 50)
# rdd_std_mark_filter.collect()
```

```python
employee_data = [
    (1, "Arun", "Male", 24, "Chennai", 28000),
    (2, "Divya", "Female", 27, "Bengaluru", 42000),
    (3, "Kavin", "Male", 22, "Chennai", 25000),
    (4, "Meena", "Female", 30, "Hyderabad", 52000),
    (5, "Ravi", "Male", 28, "Bengaluru", 38000),
    (6, "Priya", "Female", 25, "Chennai", 46000)

]

columns = ["id", "name", "gender", "age", "city", "salary"]

new_df = spark.createDataFrame(employee_data, columns)
new_df_one = new_df.filter("salary > 40000").select("name", "city", "salary")
new_df_one.show()
```

```python
product_data = [
    (101, "Notebook", 40, 5),
    (102, "Pen", 10, 20),
    (103, "Pencil", 5, 30),
    (104, "School Bag", 650, 3),
    (105, "Eraser", 8, 15)
]

columns = ["product_id", "product_name", "price", "quantity"]

new_df_product_data = spark.createDataFrame(product_data, columns)

new_df_product_data_transformed = new_df_product_data.withColumn("total_value", (col("price") * col("quantity")))
new_df_product_data_transformed.show()
```

```python
movie_data = [
    ("Leo", "Action", 250, 7.8),
    ("Hi Nanna", "Drama", 75, 8.2),
    ("Jailer", "Action", 200, 7.1),
    ("Good Night", "Comedy", 30, 8.0),
    ("Dragon", "Comedy", 100, 7.5),
    ("Amaran", "Drama", 150, 8.3)
]

columns = ["movie_name", "category", "collection_crore", "rating"]

movie_df = spark.createDataFrame(movie_data, columns)
movie_df.filter("rating >= 8").withColumn("collection_with_10_percent_bonus", col("collection_crore") * 1.10) \
.select("movie_name", "category",  "rating", "collection_with_10_percent_bonus").show()
```
