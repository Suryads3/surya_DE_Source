```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark = SparkSession.builder.appName("pyspark-medium-partice").getOrCreate()
sc = spark.sparkContext;
employee_data = [
    (101, "Arun", "IT", 45000),
    (102, "Divya", "HR", 38000),
    (103, "Kavin", "IT", 62000),
    (104, "Meena", "Finance", 55000),
    (105, "Ravi", "IT", 29000),
    (106, "Priya", "HR", 48000),
    (107, "Sanjay", "Finance", 72000),
    (108, "Nisha", "IT", 51000),
    (109, "Vijay", "HR", 32000),
    (110, "Anu", "Finance", 42000)
]
rdd_data = sc.parallelize(employee_data)
rdd_data.collect()



```

**Question 6: RDD – Employee Salary Processing**

```python
rdd_data_filtered = rdd_data.filter(lambda filtering: (filtering[len(filtering) -2] == 'IT') & (filtering[len(filtering) -1] > 40000))
rdd_data_filtered.collect()
```

```python
rdd_data_filtered_two = rdd_data.filter(lambda filtering: (filtering[len(filtering) -2] != 'IT') & (filtering[len(filtering) -1] < 40000))
rdd_data_filtered_two.collect()
```

```python
rdd_data_filtered_two_bonus = rdd_data_filtered_two.map(lambda bonus: (bonus[len(bonus) -4], bonus[len(bonus) -3], bonus[len(bonus) -2], bonus[len(bonus) -1] * 1.10))
rdd_data_filtered_two_bonus.collect()
```

```python
rdd_data_format = rdd_data.map(lambda format: (str(format[len(format) -4]) + ' | ' +
                                               str(format[len(format) -3]) + ' | ' +
                                               str(format[len(format) -2]) + ' | ' +
                                               str(format[len(format) -1]) ))
rdd_data_format.collect()
```

**Question 7: DataFrame – Student Result Processing**

```python
student_data = [
    (1, "Arun", 85, 78, 92),
    (2, "Divya", 45, 55, 60),
    (3, "Kavin", 35, 40, 38),
    (4, "Meena", 90, 88, 95),
    (5, "Ravi", 65, 72, 58),
    (6, "Priya", 30, 85, 80),
    (7, "Sanjay", 55, 60, 65),
    (8, "Nisha", 95, 92, 89),
    (9, "Vijay", 40, 35, 70),
    (10, "Anu", 75, 80, 85)
]

columns = ["student_id", "name", "python", "sql", "pyspark"]
new_df = spark.createDataFrame(student_data, columns)
new_df.show()

```

```python
new_df_total_mark = new_df.withColumn("total_marks", (col("python") + col("sql") + col("pyspark"))).select("*")
new_df_total_mark.show()
```

```python
marks_cols = ["python", "sql", "pyspark"]
new_df_total_marks = new_df_total_mark.withColumn("average_marks",(col("total_marks") / len(marks_cols))).select("*")
new_df_total_marks.show()
```

```python
with_result_df = new_df_total_marks.withColumn("result", when((col("python") >= 40) & (col("sql") >= 40) & (col("pyspark") >= 40), "PASS").otherwise("FAIL")) \
.select("*")
with_result_df.show()
```

```python
with_grade_df = with_result_df.withColumn("grade", when(col("average_marks") >=90, "A") \
.when((col("average_marks") >=75) & (col("average_marks") < 90), "B") \
.when((col("average_marks") >=60) & (col("average_marks") < 75), "C") \
.otherwise("D")
).select("*")
with_grade_df.show()
```

```python
final_df_cols = with_grade_df.select("name", "total_marks", "average_marks", "result", "grade")
final_df_cols.show()
```

```python
final_df_cols_filtered = final_df_cols.filter((col("result") == "PASS") & (col("average_marks") >=75))
final_df_cols_filtered.show()
```
